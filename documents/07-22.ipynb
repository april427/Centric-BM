### Compare two reward function

1. $reward = \sum_{u=1}^{U} (T_u - \delta_u) + \sum_{u=1}^U \log_{1/3} \epsilon_r + \log_{1/3} \epsilon_v $ 

2. $reward = \sum_{u=1}^{U} (T_u - \delta_u) + \sum_{u=1}^U \log_{1/3} \epsilon_r + \log_{1/3} \epsilon_v + SE_u$ 

The base of logarithm, $1/3$ is chosed randomly. $\log_{1/3} \epsilon = -\frac{1}{\log 3}\log \epsilon \approx -2.5 \log\epsilon$

```python
import matplotlib.pyplot as plt
import itertools
import pandas as pd

def load_latency(log_dir, filename):
    data = pd.read_csv(log_dir + filename, skiprows=0)
    data1 = list(itertools.chain(*data.values))
    data2 = [x for x in data1 if x != -1]
    return data2

latency_trained = load_latency('RL_mimo_ofdm/PPO_2ue16Nt_logEstTh_beamAct_clutters/trained_09/', 'latency.csv')
latency_benchmark = load_latency('RL_mimo_ofdm/PPO_2ue16Nt_logEstTh_beamAct_clutters/estimated_09/', 'latency.csv')

latency_genie = load_latency('RL_mimo_ofdm/PPO_2ue16Nt_logEstTh_beamAct_clutters/genie_09/', 'latency.csv')
latency_r = load_latency('RL_mimo_ofdm/PPO_2ue16Nt_logEstTh_beamAct_clutters/other_work_09/', 'latency.csv')


log_dir_clutter = 'RL_mimo_ofdm/PPO_2ue16Nt_logEstTh_beamAct_clutters/'



latency_1TDMA = load_latency(log_dir_clutter+'1-TDMA/', 'latency.csv')
latency_7TDMA = load_latency(log_dir_clutter+'7-TDMA/', 'latency.csv')
latency_13TDMA = load_latency(log_dir_clutter+'13-TDMA/', 'latency.csv')
latency_19TDMA = load_latency(log_dir_clutter+'19-TDMA/', 'latency.csv')

latency_heuristic = load_latency(log_dir_clutter + 'heuristic_09/', 'latency.csv')


#
figure_handle = plt.figure()
figure_handle.set_size_inches(5, 3.5)
ax = figure_handle.add_subplot()

counts_trained, bins_trained, patches_trained = ax.hist(latency_trained, bins=range(7), density=True, alpha=0.7, label='PPO with clutter')
counts_benchmark, bins_benchmark, patches_benchmark = ax.hist(latency_benchmark, bins=range(7), density=True, alpha=0.7, label='EstimatedAoD with clutter')
counts_genie, bins_genie, patches_genie = ax.hist(latency_genie, bins=range(7), density=True, alpha=0.7, label='Genie with clutter')
counts_random, bins_random, patches_random = ax.hist(latency_r, bins=range(7), density=True, alpha=0.7, label='Xie[10] with clutter')

bin_centers = 0.5 * (bins_trained[:-1] + bins_trained[1:])
barwidth = 0.15
ax.cla()  # Clear the previous plot
ax.bar(bin_centers - barwidth*3/2, counts_trained, width=barwidth, align='center', alpha=0.7, label='PPO with clutter')
ax.bar(bin_centers - barwidth/2, counts_benchmark, width=barwidth, align='center', alpha=0.7, label='EstimatedAoD with clutter')
ax.bar(bin_centers + barwidth/2, counts_genie, width=barwidth, align='center', alpha=0.7, label='Genie with clutter')
ax.bar(bin_centers + barwidth*3/2, counts_random, width=barwidth, align='center', alpha=0.7, label='Xie[10] with clutter')
# ax.bar(bin_centers - barwidth*5/2, counts_trainedwo, width=barwidth, align='center', alpha=0.7, label='PPO reward without SE')

ax.grid(axis='y', linestyle='--', alpha=0.7)
ax.legend()

ax.set_xticks(bin_centers)
ax.set_xticklabels(range(6))
ax.set_xlabel('Latency')
ax.set_ylabel('Probability')
ax.tick_params(axis='both', which='major', labelsize=12)
plt.tight_layout()
plt.savefig('jfigs/latency.pdf', format='pdf', bbox_inches='tight')

####
colors2 = {
    '1tdma': '#FF9800',      # orange
    '7tdma': '#9C27B0' ,          # purple
    '13tdma': '#2196F3',      # blue
    '19tdma': '#4CAF50',    # green
    'heuristic': "#0713F4", # red
}


figure_handle = plt.figure()
figure_handle.set_size_inches(5, 3.5)
ax = figure_handle.add_subplot()
counts_1TDMA, bins_1TDMA, patches_1TDMA = ax.hist(latency_1TDMA, bins=range(7), density=True, alpha=0.7, label='1-TDMA')
counts_7TDMA, bins_7TDMA, patches_7TDMA = ax.hist(latency_7TDMA, bins=range(7), density=True, alpha=0.7, label='7-TDMA')
counts_13TDMA, bins_13TDMA, patches_13TDMA = ax.hist(latency_13TDMA, bins=range(7), density=True, alpha=0.7, label='13-TDMA')
counts_19TDMA, bins_19TDMA, patches_19TDMA = ax.hist(latency_19TDMA, bins=range(7), density=True, alpha=0.7, label='19-TDMA')
counts_heuristic, bins_heuristic, patches_heuristic = ax.hist(latency_heuristic, bins=range(7), density=True, alpha=0.7, label='Heuristic')

bin_centers = 0.5 * (bins_1TDMA[:-1] + bins_1TDMA[1:])
barwidth = 0.15
ax.cla()  # Clear the previous plot
ax.bar(bin_centers - barwidth*3/2, counts_1TDMA, width=barwidth, align='center', alpha=0.7, label='1-TDMA with clutter')
ax.bar(bin_centers - barwidth/2, counts_7TDMA, width=barwidth, align='center', alpha=0.7, label='7-TDMA with clutter')
ax.bar(bin_centers + barwidth/2, counts_13TDMA, width=barwidth, align='center', alpha=0.7, label='13-TDMA with clutter')
ax.bar(bin_centers + barwidth*3/2, counts_19TDMA, width=barwidth, align='center', alpha=0.7, label='19-TDMA with clutter')
ax.bar(bin_centers + barwidth*5/2, counts_trained, width=barwidth, align='center', alpha=0.7, label='PPO with clutter')
ax.bar(bin_centers + barwidth*7/2, counts_heuristic, width=barwidth, align='center', alpha=0.7, label='Heuristic with clutter')

ax.set_xticks(bin_centers)
ax.set_xticklabels(range(6))
ax.tick_params(axis='both', which='major', labelsize=12)
ax.set_xlabel('Latency')
ax.set_ylabel('Probability')
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig('jfigs/latency_tdma.pdf', format='pdf', bbox_inches='tight')
plt.show()
```

2. User-beam association: AoD estimation methods, such as MUSIC, achieve decent performance. They nevertheless suffer from the user-beam association problem when there are multiple users. The association information is acknowledged to the BS via user feedback channel after the users obtained the optimal one from the BS sweept beams. For the trained PPO method, the agent acquires such information based on the previous allocation beams which is input as part of the state. In addition, the AoD-estimated method is expected to have degraded performance when there are clutters in the environment as the they contribute to the refelcted echoes. The BS requires further information to distinguish AoDs of users from these of clutters. On contrary, the PPO agent learns the environment implicitly via the network. For a fair comparison, we consider that the BS has the user-beam association information while adopting the benchmark: AoD-estimated method.
